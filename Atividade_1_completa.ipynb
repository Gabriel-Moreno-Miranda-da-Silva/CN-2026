### Parte A – truncamento e arredondamento

```python
np.set_printoptions(suppress=True)
x1 = 3.14159265
x2 = 98.76543
x3 = 0.00498765

Lista_Casas = [0,1,2,3,4]
Lista_Arred = np.zeros((5,3))
Lista_trunc = np.zeros((5,3))

def truncar(numero, casas):
  fator = 10 ** casas
  return math.trunc(numero * fator) / fator

for i in Lista_Casas:
  Lista_Arred[i][0] = round(x1,Lista_Casas[i])
  Lista_Arred[i][1] = round(x2,Lista_Casas[i])
  Lista_Arred[i][2] = round(x3,Lista_Casas[i])
  Lista_trunc[i][0] = truncar(x1,Lista_Casas[i])
  Lista_trunc[i][1] = truncar(x2,Lista_Casas[i])
  Lista_trunc[i][2] = truncar(x3,Lista_Casas[i])

def erro_absoluto(valor_referencia, valor_aproximado):
  return abs(valor_referencia - valor_aproximado)

def erro_relativo(valor_referencia, valor_aproximado):
  return abs(valor_referencia - valor_aproximado) / abs(valor_referencia)

def erro_percentual(valor_referencia, valor_aproximado):
  return 100.0 * abs(valor_referencia - valor_aproximado) / abs(valor_referencia)

Lista_Arred_errA = np.zeros((5,3))
Lista_Arred_errR = np.zeros((5,3))
Lista_Arred_errP = np.zeros((5,3))
Lista_trunc_errA = np.zeros((5,3))
Lista_trunc_errR = np.zeros((5,3))
Lista_trunc_errP = np.zeros((5,3))

for i in Lista_Casas:
  Lista_Arred_errA[i][0] = erro_absoluto(x1, Lista_Arred[i][0])
  Lista_Arred_errA[i][1] = erro_absoluto(x2, Lista_Arred[i][1])
  Lista_Arred_errA[i][2] = erro_absoluto(x3, Lista_Arred[i][2])
  Lista_trunc_errA[i][0] = erro_absoluto(x1, Lista_trunc[i][0])
  Lista_trunc_errA[i][1] = erro_absoluto(x2, Lista_trunc[i][1])
  Lista_trunc_errA[i][2] = erro_absoluto(x3, Lista_trunc[i][2])
  Lista_Arred_errR[i][0] = erro_relativo(x1, Lista_Arred[i][0])
  Lista_Arred_errR[i][1] = erro_relativo(x2, Lista_Arred[i][1])
  Lista_Arred_errR[i][2] = erro_relativo(x3, Lista_Arred[i][2])
  Lista_trunc_errR[i][0] = erro_relativo(x1, Lista_trunc[i][0])
  Lista_trunc_errR[i][1] = erro_relativo(x2, Lista_trunc[i][1])
  Lista_trunc_errR[i][2] = erro_relativo(x3, Lista_trunc[i][2])
  Lista_Arred_errP[i][0] = erro_percentual(x1, Lista_Arred[i][0])
  Lista_Arred_errP[i][1] = erro_percentual(x2, Lista_Arred[i][1])
  Lista_Arred_errP[i][2] = erro_percentual(x3, Lista_Arred[i][2])
  Lista_trunc_errP[i][0] = erro_percentual(x1, Lista_trunc[i][0])
  Lista_trunc_errP[i][1] = erro_percentual(x2, Lista_trunc[i][1])
  Lista_trunc_errP[i][2] = erro_percentual(x3, Lista_trunc[i][2])

print("Lista Arredondada:\n", Lista_Arred)
print("Lista Truncada:\n", Lista_trunc)
print("Erro Absoluto, Lista Arredondada:\n", Lista_Arred_errA)
print("Erro Absoluto, Lista Truncada:\n", Lista_trunc_errA)
print("Erro Relativo, Lista Arredondada:\n", Lista_Arred_errR)
print("Erro Relativo, Lista Truncada:\n", Lista_trunc_errR)
print("Erro Percentual, Lista Arredondada:\n", Lista_Arred_errP)
print("Erro Percentual, Lista Truncada:\n", Lista_trunc_errP)

cores = ['red', 'green', 'blue']
x_labels = ['X1', 'X2', 'X3']

plt.figure(figsize=(6, 4))
for i in range(Lista_trunc_errA.shape[1]):
  plt.plot(
      Lista_Casas,
      Lista_Arred_errA[:, i],
      marker='o',
      color=cores[i],
      label=x_labels[i],
      )
plt.yscale('log')
plt.xlabel('Casas Decimais')
plt.ylabel('Erro Absoluto (escala logarítmica)')
plt.title('Gráfico de Erro Absoluto em Arredondamento')
plt.grid(True, which='both', ls='--')
plt.legend()
plt.show()

plt.figure(figsize=(6, 4))
for i in range(Lista_trunc_errA.shape[1]):
  plt.plot(
      Lista_Casas,
      Lista_trunc_errA[:, i],
      marker='o',
      color=cores[i],
      label=x_labels[i],
      )
plt.yscale('log')
plt.xlabel('Casas Decimais')
plt.ylabel('Erro Absoluto (escala logarítmica)')
plt.title('Gráfico de Erro Absoluto em Truncamento')
plt.grid(True, which='both', ls='--')
plt.legend()
plt.show()
```

```python
def media (valores):
  resultado = 0;
  for i, VALOR in enumerate(valores):
    resultado += valores[i]
  return resultado / len(valores)

def minimo (valores):
  return min(valores)

def maximo (valores):
  return max(valores)

def amplitude (valores):
  return max(valores) - min(valores)

def desvio_padrao (valores):
  return statistics.stdev(valores)

def truncar(numero, casas):
  fator = 10 ** casas
  return math.trunc(numero * fator) / fator

Lista_Pressao = [121.7, 120.4, 122.1, 119.8, 121.2, 120.9, 122.4, 121.0, 120.2, 121.5]
Lista_Pressao_arred_int = [round(n) for n in Lista_Pressao]
Lista_Pressao_trunc_int = [truncar(n, 0) for n in Lista_Pressao]

print("Dados Originais:")
print(" Média das Pressões:",media(Lista_Pressao))
print(" Mínimo das Pressões:",minimo(Lista_Pressao))
print(" Máximo das Pressões:",maximo(Lista_Pressao))
print(" Amplitude das Pressões:",amplitude(Lista_Pressao))
print(" Desvio Padrão das Pressões:",desvio_padrao(Lista_Pressao))
print("-----------------------------")
print("Dados Arredondados:")
print(" Média das Pressões:",media(Lista_Pressao_arred_int))
print(" Mínimo das Pressões:",minimo(Lista_Pressao_arred_int))
print(" Máximo das Pressões:",maximo(Lista_Pressao_arred_int))
print(" Amplitude das Pressões:",amplitude(Lista_Pressao_arred_int))
print(" Desvio Padrão das Pressões:",desvio_padrao(Lista_Pressao_arred_int))
print("-----------------------------")
print("Dados Truncados:")
print(" Média das Pressões:",media(Lista_Pressao_trunc_int))
print(" Mínimo das Pressões:",minimo(Lista_Pressao_trunc_int))
print(" Máximo das Pressões:",maximo(Lista_Pressao_trunc_int))
print(" Amplitude das Pressões:",amplitude(Lista_Pressao_trunc_int))
print(" Desvio Padrão das Pressões:",desvio_padrao(Lista_Pressao_trunc_int))
```

# C.1 Precisão das variáveis


```python
import math
import pandas as pd


def hagen(r, l, u, p):
  pi = math.pi
  return (pi * (r**4) * p) / (8 * l * u)


def truncar(numero, casas):
  fator = 10**casas
  return math.trunc(numero * fator) / fator


def erro_absoluto(ref, aprox):
  return abs(ref - aprox)


def erro_relativo(ref, aprox):
  return erro_absoluto(ref, aprox) / abs(ref)



r_mm = 0.8
r = r_mm * 1e-3
L = 0.2
u = 3.5e-3
p = 1200.0


Q_referencia = hagen(r, L, u, p)
print("Q de referencia", Q_referencia)
print("-----------------------")

r_trunc = truncar(r_mm, 1) * 1e-3
r_aprox = round(r_mm, 1) * 1e-3
Q_rtrunc = hagen(r_trunc, L, u, p)
Q_raprox = hagen(r_aprox, L, u, p)

u_trunc = truncar(u, 3)
u_aprox = round(u, 3)
Q_utrunc = hagen(r, L, u_trunc, p)
Q_uaprox = hagen(r, L, u_aprox, p)

p_trunc = truncar(p / 100.0, 0) * 100.0
p_aprox = round(p / 100.0) * 100.0
Q_ptrunc = hagen(r, L, u, p_trunc)
Q_paprox = hagen(r, L, u, p_aprox)

Q_trunc = hagen(r_trunc, L, u_trunc, p_trunc)
Q_aprox = hagen(r_aprox, L, u_aprox, p_aprox)

cenarios = [
    ("1. ", Q_referencia, Q_referencia),
    ("2. ", Q_rtrunc, Q_raprox),
    ("3. ", Q_utrunc, Q_uaprox),
    ("4. ", Q_ptrunc, Q_paprox),
    ("5. ", Q_trunc, Q_aprox),
]

dados = []
for nome, q_tr, q_ap in cenarios:
    e_abs_tr = erro_absoluto(Q_referencia, q_tr)
    e_rel_tr = erro_relativo(Q_referencia, q_tr)

    e_abs_ap = erro_absoluto(Q_referencia, q_ap)
    e_rel_ap = erro_relativo(Q_referencia, q_ap)

    dados.append({
        "Cenário": nome,
        "Q Truncado(m³/s)": f"{q_tr:.6e}",
        "Erro Abs(trunc)": f"{e_abs_tr:.6e}",
        "Erro Rel(trunc)": f"{e_rel_tr * 100:.4f}%",
        "Q Arredondado(m³/s)": f"{q_ap:.6e}",
        "Erro Abs(aprox)": f"{e_abs_ap:.6e}",
        "Erro Rel(aprox)": f"{e_rel_ap * 100:.4f}%",
    })

df = pd.DataFrame(dados)
print(df.to_string(index=False))
```

# C.2 Sensibilidade

```python
def hagen(r, L, u, p):
    return (math.pi * (r ** 4) * p) / (8 * u * L)

def truncar(valor, casas):
    fator = 10 ** casas
    return math.trunc(valor * fator) / fator

def erro_absoluto(ref, aprox):
    return abs(ref - aprox)

def erro_relativo(ref, aprox):
    return abs(ref - aprox) / abs(ref)

L = 0.20
u = 3.5e-3
p = 1200.0

lista_r_mm = []
lista_Q_ref = []
lista_Q_trunc = []
lista_Q_aprox = []
lista_ea_trunc = []
lista_ea_aprox = []
lista_er_trunc = []
lista_er_aprox = []
lista_ep_trunc = []
lista_ep_aprox = []

for i in range(70, 91):
    r_mm = i / 100.0
    r_orig = r_mm * 1e-3

    r_trunc_mm = truncar(r_mm, 1)
    r_arred_mm = round(r_mm, 1)

    r_trunc = r_trunc_mm * 1e-3
    r_arred = r_arred_mm * 1e-3

    Q_ref = hagen(r_orig, L, u, p)
    Q_trunc = hagen(r_trunc, L, u, p)
    Q_arred = hagen(r_arred, L, u, p)

    e_t = erro_absoluto(Q_ref, Q_trunc)
    e_a = erro_absoluto(Q_ref, Q_arred)

    er_t = erro_relativo(Q_ref, Q_trunc)
    er_a = erro_relativo(Q_ref, Q_arred)

    ep_t = 100 * er_t
    ep_a = 100 * er_a

    lista_r_mm.append(r_mm)
    lista_Q_ref.append(Q_ref)
    lista_Q_trunc.append(Q_trunc)
    lista_Q_aprox.append(Q_arred)
    lista_ea_trunc.append(e_t)
    lista_ea_aprox.append(e_a)
    lista_er_trunc.append(er_t)
    lista_er_aprox.append(er_a)
    lista_ep_trunc.append(ep_t)
    lista_ep_aprox.append(ep_a)

r_suave_mm = np.linspace(0.70, 0.90, 500)
Q_suave_ref = [hagen(r * 1e-3, L, u, p) for r in r_suave_mm]

fig, axes = plt.subplots(2, 2, figsize=(12, 10))

# primeiro gráfico
axes[0, 0].plot(r_suave_mm, Q_suave_ref, label=r'$Q$ Referência (exato)', color='black', linewidth=2)
axes[0, 0].step(lista_r_mm, lista_Q_trunc, label=r'$Q$ com $r$ Truncado', color='tab:red', linestyle='--', where='mid')
axes[0, 0].step(lista_r_mm, lista_Q_aprox, label=r'$Q$ com $r$ Arredondado', color='tab:blue', linestyle=':', where='mid')
axes[0, 0].set_title(r'Vazão $Q$ em função do Raio $r$', fontsize=12)
axes[0, 0].set_xlabel(r'Raio $r$ (mm)', fontsize=10)
axes[0, 0].set_ylabel(r'Vazão $Q$ (m$^3$/s)', fontsize=10)
axes[0, 0].grid(True, linestyle='--', alpha=0.6)
axes[0, 0].legend(fontsize=9)
axes[0, 0].ticklabel_format(style='sci', axis='y', scilimits=(0,0))

# segundo gráfico
axes[0, 1].plot(lista_r_mm, lista_ea_trunc, label='Erro Absoluto (Truncamento)', color='tab:red', marker='o', linestyle='--')
axes[0, 1].plot(lista_r_mm, lista_ea_aprox, label='Erro Absoluto (Arredondamento)', color='tab:blue', marker='s', linestyle=':')
axes[0, 1].set_title(r'Erro Absoluto de $Q$ em função do Raio $r$', fontsize=12)
axes[0, 1].set_xlabel(r'Raio $r$ (mm)', fontsize=10)
axes[0, 1].set_ylabel(r'Erro Absoluto $|Q_{ref} - Q_{aprox}|$ (m$^3$/s)', fontsize=10)
axes[0, 1].grid(True, linestyle='--', alpha=0.6)
axes[0, 1].legend(fontsize=9)
axes[0, 1].ticklabel_format(style='sci', axis='y', scilimits=(0,0))

# terceiro gráfico
axes[1, 0].plot(lista_r_mm, lista_er_trunc, label='Erro Relativo (Truncamento)', color='tab:red', marker='o', linestyle='--')
axes[1, 0].plot(lista_r_mm, lista_er_aprox, label='Erro Relativo (Arredondamento)', color='tab:blue', marker='s', linestyle=':')
axes[1, 0].set_title(r'Erro Relativo de $Q$ em função do Raio $r$', fontsize=12)
axes[1, 0].set_xlabel(r'Raio $r$ (mm)', fontsize=10)
axes[1, 0].set_ylabel(r'Erro Relativo', fontsize=10)
axes[1, 0].grid(True, linestyle='--', alpha=0.6)
axes[1, 0].legend(fontsize=9)

# quarto gráfico
axes[1, 1].plot(lista_r_mm, lista_ep_trunc, label='Erro Percentual (Truncamento)', color='tab:red', marker='o', linestyle='--')
axes[1, 1].plot(lista_r_mm, lista_ep_aprox, label='Erro Percentual (Arredondamento)', color='tab:blue', marker='s', linestyle=':')
axes[1, 1].set_title(r'Erro Percentual de $Q$ em função do Raio $r$', fontsize=12)
axes[1, 1].set_xlabel(r'Raio $r$ (mm)', fontsize=10)
axes[1, 1].set_ylabel(r'Erro Percentual $E_p$ (%)', fontsize=10)
axes[1, 1].grid(True, linestyle='--', alpha=0.6)
axes[1, 1].legend(fontsize=9)


plt.tight_layout()
plt.show()
```

# D.1 Erro geométrico

```python
from enum import verify
def hagen(r, L, u, p):
    return (math.pi * (r ** 4) * p) / (8 * u * L)

def erro_absoluto(ref, aprox):
    return abs(ref - aprox)

def erro_relativo(ref, aprox):
    return abs(ref - aprox) / abs(ref)

def r_negativo(r, vr):
    return r - vr

def r_positivo(r, vr):
    return r + vr

r = 50.0 * 1e-9
L = 10 * 1e-6
u = 3.5e-3
p = 5000
Q_ref = hagen(r, L, u, p)
print("Vazão nocatetar", Q_ref)

lista_Q_ref = []
lista_dr_nm = []
lista_Q_menos = []
lista_Q_mais = []
lista_ea_max = []
lista_er_max = []
lista_ep_max = []
lista_er_linear = []

for vr in range(1, 51):
    deltar_nm = vr / 10.0
    deltar = deltar_nm * 1e-9

    rn = r_negativo(r, deltar)
    rp = r_positivo(r, deltar)

    Q_rnegativo = hagen(rn, L, u, p)
    Q_rpositivo = hagen(rp, L, u, p)

    ea_t = erro_absoluto(Q_ref, Q_rnegativo)
    ea_a = erro_absoluto(Q_ref, Q_rpositivo)

    er_t = erro_relativo(Q_ref, Q_rnegativo)
    er_a = erro_relativo(Q_ref, Q_rpositivo)

    ea_max = max(ea_t, ea_a)
    er_max = max(er_t, er_a)
    ep_max = er_max * 100.0

    er_linear = 4.0 * (deltar / r)

    lista_Q_ref.append(Q_ref)
    lista_dr_nm.append(deltar_nm)
    lista_Q_menos.append(Q_rnegativo)
    lista_Q_mais.append(Q_rpositivo)
    lista_ea_max.append(ea_max)
    lista_er_max.append(er_max)
    lista_ep_max.append(ep_max)
    lista_er_linear.append(er_linear)


fig, axes = plt.subplots(2, 2, figsize=(12, 10))

# primeiro gráfico
axes[0, 0].plot(lista_dr_nm, lista_Q_mais, label=r'$Q(r_+) = Q(r + \Delta r)$', color='tab:blue', linestyle='--', linewidth=1.8)
axes[0, 0].plot(lista_dr_nm, lista_Q_ref, label=r'$Q(r)$ Referência', color='black', linewidth=2.0)
axes[0, 0].plot(lista_dr_nm, lista_Q_menos, label=r'$Q(r_-) = Q(r - \Delta r)$', color='tab:red', linestyle=':', linewidth=1.8)
axes[0, 0].set_title(r'Vazão $Q$ para Limites do Raio ($r_-$ e $r_+$)', fontsize=11)
axes[0, 0].set_xlabel(r'Incerteza do Raio $\Delta r$ (nm)', fontsize=10)
axes[0, 0].set_ylabel(r'Vazão $Q$ (m$^3$/s)', fontsize=10)
axes[0, 0].grid(True, linestyle='--', alpha=0.6)
axes[0, 0].legend(fontsize=9)
axes[0, 0].ticklabel_format(style='sci', axis='y', scilimits=(0,0))


# segundo gráfico

axes[0, 1].plot(lista_dr_nm, lista_ea_max, color='tab:red', marker='o', markersize=3, linestyle='--', label='Erro Absoluto Máximo')
axes[0, 1].set_title(r'Erro Absoluto Máximo de $Q$ em função de $\Delta r$', fontsize=11)
axes[0, 1].set_xlabel(r'Incerteza do Raio $\Delta r$ (nm)', fontsize=10)
axes[0, 1].set_ylabel(r'Erro Absoluto Máximo (m$^3$/s)', fontsize=10)
axes[0, 1].grid(True, linestyle='--', alpha=0.6)
axes[0, 1].legend(fontsize=9)
axes[0, 1].ticklabel_format(style='sci', axis='y', scilimits=(0,0))

# terceiro gráfico
axes[1, 0].plot(lista_dr_nm, lista_er_max, color='tab:blue', marker='s', markersize=3, linestyle=':', label='Erro Relativo Máximo')
axes[1, 0].set_title(r'Erro Relativo Máximo de $Q$ em função de $\Delta r$', fontsize=11)
axes[1, 0].set_xlabel(r'Incerteza do Raio $\Delta r$ (nm)', fontsize=10)
axes[1, 0].set_ylabel(r'Erro Relativo Máximo', fontsize=10)
axes[1, 0].grid(True, linestyle='--', alpha=0.6)
axes[1, 0].legend(fontsize=9)

# quarto gráfico
axes[1, 1].plot(lista_dr_nm, lista_ep_max, color='tab:green', marker='^', markersize=3, linestyle=':', label='Erro Porcentual Máximo')
axes[1, 1].set_title(r'Erro Porcentual Máximo de $Q$ em função de $\Delta r$', fontsize=11)
axes[1, 1].set_xlabel(r'Incerteza do Raio $\Delta r$ (nm)', fontsize=10)
axes[1, 1].set_ylabel(r'Erro Porcentual Máximo (%)', fontsize=10)
axes[1, 1].grid(True, linestyle='--', alpha=0.6)
axes[1, 1].legend(fontsize=9)

plt.tight_layout()
plt.show()
```

# D.2 Precisão computacional


```python
def hagen(r, L, u, p, dtipo):
    r2 = dtipo(r)
    L2 = dtipo(L)
    u2 = dtipo(u)
    p2 = dtipo(p)
    pi2 = dtipo(np.pi)
    return (pi2 * (r2 ** dtipo(4)) * p2) / (dtipo(8) * u2 * L2)

def r_positivo(r, vr):
  return r+vr

r=50.0 * 1e-9
L=10.0 * 1e-6
u=3.5e-3
p=5000


r4_32 = np.float32(r) ** np.float32(4)
Q_32 = hagen(r, L, u, p, np.float32)

r4_64 = np.float64(r) ** np.float64(4)
Q_64 = hagen(r, L, u, p, np.float64)

dif_a_r = abs(r4_64 - r4_32)
dif_r_r = dif_a_r / abs(r4_64)

dif_a_Q = abs(Q_64 - Q_32)
dif_r_Q = dif_a_Q / abs(Q_64)

eps_32 = np.finfo(np.float32).eps
eps_64 = np.finfo(np.float64).eps

Q_ref = Q_64

print(f"Epsilon float32  : {eps_32:.6e}")
print(f"Epsilon float64  : {eps_64:.6e}")
print(f"r^4 (float32)    : {r4_32:.10e}")
print(f"r^4 (float64)    : {r4_64:.10e}")
print(f"Erro Relativo r^4: {dif_r_r:.6e}")
print(f"Q (float32)      : {Q_32:.10e}")
print(f"Q (float64)      : {Q_64:.10e}")
print(f"Erro Relativo Q  : {dif_r_Q:.6e}\n")

dados_delta = []
diferenca = 1.0
for i in range(1, 16):
    vr_r = 10.0 ** (-i)
    vr = vr_r * r
    rp = r_positivo(r, vr)
    Q_rpositivo = hagen(rp, L, u, p, np.float64)

    delta1 = (Q_rpositivo - Q_ref) / Q_ref


    delta2 = ((1.0 + vr_r) ** 4) - 1.0

    diferenca = abs(delta1 - delta2)
    dados_delta.append({
      "i": i,
      "Δr / r": f"1e-{i}",
      "δ1 (via Q)": f"{delta1:.8e}",
      "δ2 (expandido)": f"{delta2:.8e}",
      "|δ1 - δ2|": f"{diferenca:.8e}",
    })
print("---------------------")
print("diferença absoluta", diferenca)
print("---------------------")
df_delta = pd.DataFrame(dados_delta)
print(df_delta.to_string(index=False))

```

Conforme Δr/r vai progredindo, o primeiro método irá cancelar termos importantes pois Q(r+) e Q(r) se tornam quase identicos, o que gera perda de significância
Já o segundo método calcula de forma mais estável e evita os cancelamentos de termos próximos para valores intermediário de Δr
Além disso, quando Δr/r<=10^-14, o float.64 não consegue mais distinguir os termos, zerando a diferença e os dois métodos colapsam para zero

```python

from matplotlib.animation import FuncAnimation
from IPython.display import Image, display

# -------------------------------------------------------------------------
# 1. PARÂMETROS E FUNÇÕES FÍSICAS (PARTE D)
# -------------------------------------------------------------------------
L = 10.0e-6        # Comprimento: 10 µm em m
u = 3.5e-3         # Viscosidade: 3.5e-3 Pa.s
p = 5000.0         # Diferença de pressão: 5000 Pa
r_ref_nm = 50.0    # Raio de referência: 50 nm

def hagen_poiseuille(r_m):
    """Calcula a vazão Q (m³/s) para o raio em metros."""
    return (math.pi * (r_m ** 4) * p) / (8 * u * L)

# Raio de referência e Vazão de referência
Q_ref = hagen_poiseuille(r_ref_nm * 1e-9)

# -------------------------------------------------------------------------
# 2. CONFIGURAÇÃO DOS 30 QUADROS (45 nm a 55 nm)
# -------------------------------------------------------------------------
num_frames = 30
r_array_nm = np.linspace(45.0, 55.0, num_frames)
r_array_m = r_array_nm * 1e-9
Q_array = [hagen_poiseuille(r) for r in r_array_m]

# -------------------------------------------------------------------------
# 3. CRIAÇÃO DA FIGURA COM 2 SUBPLOTS
# -------------------------------------------------------------------------
fig, (ax_canal, ax_curva) = plt.subplots(1, 2, figsize=(12, 5), dpi=100)
fig.patch.set_facecolor('#121214')

for ax in [ax_canal, ax_curva]:
    ax.set_facecolor('#1e1e24')
    ax.tick_params(colors='white')
    ax.xaxis.label.set_color('white')
    ax.yaxis.label.set_color('white')
    ax.title.set_color('#00e676')
    for spine in ax.spines.values():
        spine.set_color('#323238')

# -------------------------------------------------------------------------
# 4. FUNÇÃO DE ATUALIZAÇÃO DO GIF (QUADRO A QUADRO)
# -------------------------------------------------------------------------
def update(frame):
    ax_canal.clear()
    ax_curva.clear()

    # Re-aplica estilo dark após clear()
    for ax in [ax_canal, ax_curva]:
        ax.set_facecolor('#1e1e24')
        ax.tick_params(colors='white')
        ax.xaxis.label.set_color('white')
        ax.yaxis.label.set_color('white')
        ax.title.set_color('#00e676')
        for spine in ax.spines.values():
            spine.set_color('#323238')

    # Dados do quadro atual
    r_atual_nm = r_array_nm[frame]
    Q_atual = Q_array[frame]

    # Erro percentual do raio e da vazão em relação a r = 50 nm
    erro_p_r = (abs(r_atual_nm - r_ref_nm) / r_ref_nm) * 100
    erro_p_Q = (abs(Q_atual - Q_ref) / Q_ref) * 100

    # ---------------------------------------------------------------------
    # PAINEL ESQUERDO: Seção Transversal do Canal
    # ---------------------------------------------------------------------
    circulo_parede = plt.Circle((0, 0), 60, color='#323238', fill=True)
    circulo_fluido = plt.Circle((0, 0), r_atual_nm, color='#00e676', alpha=0.7)
    circulo_ref = plt.Circle((0, 0), r_ref_nm, color='white', fill=False, linestyle='--', linewidth=1.5, label='Ref (50 nm)')

    ax_canal.add_patch(circulo_parede)
    ax_canal.add_patch(circulo_fluido)
    ax_canal.add_patch(circulo_ref)
    ax_canal.set_xlim(-65, 65)
    ax_canal.set_ylim(-65, 65)
    ax_canal.set_aspect('equal')
    ax_canal.set_title('Seção Transversal do Nanocateter', fontsize=12, fontweight='bold')
    ax_canal.set_xlabel('nm')
    ax_canal.set_ylabel('nm')
    ax_canal.legend(loc='upper right', facecolor='#121214', edgecolor='#323238', labelcolor='white', fontsize=8)

    # Painel com as informações de texto
    info_text = (
        f"Raio Atual (r): {r_atual_nm:.2f} nm\n"
        f"Vazão (Q): {Q_atual:.4e} m³/s\n"
        f"Erro % em r: {erro_p_r:.2f}%\n"
        f"Erro % em Q: {erro_p_Q:.2f}%"
    )
    ax_canal.text(-60, -58, info_text, color='white', fontsize=9,
                  bbox=dict(boxstyle='round,pad=0.5', facecolor='#121214', edgecolor='#00e676', alpha=0.9))

    # ---------------------------------------------------------------------
    # PAINEL DIREITO: Curva Q x r Construída Progressivamente
    # ---------------------------------------------------------------------
    # Curva sendo construída do início até o quadro atual
    ax_curva.plot(r_array_nm[:frame+1], Q_array[:frame+1], color='#00e676', linewidth=2.5, label='Curva Q(r)')
    # Ponto atual destacado
    ax_curva.plot(r_atual_nm, Q_atual, marker='o', markersize=8, color='tab:red', label='Ponto Atual')
    # Ponto de referência fixo
    ax_curva.plot(r_ref_nm, Q_ref, marker='x', markersize=8, color='white', label='Ref (50 nm)')

    ax_curva.set_xlim(44, 56)
    ax_curva.set_ylim(min(Q_array)*0.9, max(Q_array)*1.1)
    ax_curva.set_title('Construção da Curva Q vs r', fontsize=12, fontweight='bold')
    ax_curva.set_xlabel('Raio r (nm)')
    ax_curva.set_ylabel('Vazão Q (m³/s)')
    ax_curva.grid(True, linestyle='--', alpha=0.3)
    ax_curva.ticklabel_format(style='sci', axis='y', scilimits=(0,0))
    ax_curva.legend(loc='upper left', facecolor='#121214', edgecolor='#323238', labelcolor='white', fontsize=8)

plt.tight_layout()

# -------------------------------------------------------------------------
# 5. GERAÇÃO E SALVAMENTO DO GIF (Pillow)
# -------------------------------------------------------------------------
anim = FuncAnimation(fig, update, frames=num_frames, interval=150)
gif_filename = 'nanocateter_animacao.gif'
anim.save(gif_filename, writer='pillow', fps=7)
plt.close()

# -------------------------------------------------------------------------
# 6. EXIBIÇÃO DO GIF NO GOOGLE COLAB
# -------------------------------------------------------------------------
print("GIF gerado com sucesso!")
display(Image(open(gif_filename, 'rb').read()))
```
